```python
import numpy as np
from numpy.typing import NDArray
import time
import tensorflow as tf
```

```python
# Generate Dummy Data

data_size = 100000000

num_true_positive = np.random.randint(0, 100000 // 2, data_size)
num_false_positive = np.random.randint(0, 100000 // 2, data_size)
num_true_negative = np.random.randint(0, 100000 // 2, data_size)
num_false_negative = np.random.randint(0, 100000 // 2, data_size)
sample_size = num_true_negative + num_false_positive + num_true_negative + num_false_negative
print(sample_size.shape)
```

```python
# With loops

def f1_sequential(true_positives: NDArray, false_positives: NDArray, false_negatives: NDArray) -> NDArray:
    assert len(true_positives) == len(false_positives) == len(false_negatives)
    return_array = np.zeros(len(true_positives))
    for i in range(len(true_positives)):
        return_array[i] = 2 * true_positives[i] / (2 * true_positives[i] + false_positives[i] + false_negatives[i])
    return return_array

start_time = time.time()
results = f1_sequential(num_true_positive, num_false_positive, num_false_negative)
end_time = time.time()
print(f"Execution time: {(end_time - start_time):.4} s")
```

```python
def f1_parallelized_numpy(true_positives: NDArray, false_positives: NDArray, false_negatives: NDArray) -> NDArray:
    assert len(true_positives) == len(false_positives) == len(false_negatives)
    return 2*true_positives / (2*true_positives + false_positives + false_negatives)

start_time = time.time()
results = f1_parallelized_numpy(num_true_positive, num_false_positive, num_false_negative)
end_time = time.time()
print(f"Execution time: {(end_time - start_time)} s")
```

```python
def f1_parallelized_tensorflow(true_positives: NDArray, false_positives: NDArray, false_negatives: NDArray) -> NDArray:
    assert len(true_positives) == len(false_positives) == len(false_negatives)

    double_true_positives = tf.multiply(tf.cast(true_positives, tf.float32), 2)
    false_positives = tf.cast(false_positives, tf.float32)
    false_negatives = tf.cast(false_negatives, tf.float32)

    return tf.divide(double_true_positives, tf.add(tf.add(double_true_positives, false_positives), false_negatives))

start_time = time.time()
results = f1_parallelized_tensorflow(num_true_positive, num_false_positive, num_false_negative)
end_time = time.time()
print(f"Execution time: {(end_time - start_time)} s")
```
